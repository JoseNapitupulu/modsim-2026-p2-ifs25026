## Q1.Pada kusioner dari keseluruhan data, skala mana yang paling banyak dipilih oleh partisipan? sertakan jumlah dan persen

```python
import pandas as pd
import plotly.express as px

def plot_from_print(judul, hasil_print, label_x):
    skala, jumlah, persen = hasil_print.split("|")

    df_plot = pd.DataFrame({
        label_x: [skala],
        "Jumlah": [int(jumlah)],
        "Persen": [float(persen)]
    })

    fig = px.bar(
        df_plot,
        x=label_x,
        y="Jumlah",
        text="Persen",
        title=judul
    )

    fig.update_traces(texttemplate='%{text}%')
    fig.show()

plot_from_print(
    "Q1 - Skala Paling Banyak Dipilih",
    "CS|500|70.1",
    "Skala"
)

```

## Q9 Pada kusioner dari pertanyaan Q1 sampai Q17, pertanyaan mana saja yang terdapat memilih skala STS (Sangat Tidak Setuju)? sertakan jumlah dan persen
    

```python
import pandas as pd
import plotly.express as px

def plot_from_print(judul, hasil_print, label_x):
    skala, jumlah, persen = hasil_print.split("|")

    df_plot = pd.DataFrame({
        label_x: [skala],
        "Jumlah": [int(jumlah)],
        "Persen": [float(persen)]
    })

    fig = px.bar(
        df_plot,
        x=label_x,
        y="Jumlah",
        text="Persen",
        title=judul
    )

    fig.update_traces(texttemplate='%{text}%')
    fig.show()

hasil_q9 = "Q1:0.9|Q2:0.9|Q9:0.9|Q11:0.9"

rows = []
for item in hasil_q9.split("|"):
    pertanyaan, persen = item.split(":")
    rows.append([pertanyaan, float(persen)])

df_q9 = pd.DataFrame(rows, columns=["Pertanyaan","Persen"])

fig = px.bar(
    df_q9,
    x="Pertanyaan",
    y="Persen",
    text="Persen",
    title="Q9 - Pertanyaan dengan Jawaban STS"
)

fig.update_traces(texttemplate='%{text}%')

fig.show()
```

## Q10 Jika skala skor untuk masing masing skala adalah 6, 5, 4, 3, 2, 1 untuk SS, S, CS, CTS, TS, STS Berapa skor rata-rata keseluruhan pertanyaan?

```python
import pandas as pd
import plotly.express as px

def plot_from_print(judul, hasil_print, label_x):
    skala, jumlah, persen = hasil_print.split("|")

    df_plot = pd.DataFrame({
        label_x: [skala],
        "Jumlah": [int(jumlah)],
        "Persen": [float(persen)]
    })

    fig = px.bar(
        df_plot,
        x=label_x,
        y="Jumlah",
        text="Persen",
        title=judul
    )

    fig.update_traces(texttemplate='%{text}%')
    fig.show()

df_q10 = pd.DataFrame({
    "Metric":["Rata-rata"],
    "Nilai":[4.80]
})

fig = px.bar(df_q10, x="Metric", y="Nilai",
             text="Nilai",
             title="Q10 - Rata-rata Skor")

fig.show()
```

## Q11 Jika skala skor untuk masing masing skala adalah 6, 5, 4, 3, 2, 1 untuk SS, S, CS, CTS, TS, STS Maka pada kusioner dari pertanyaan Q1 sampai Q17, pertanyaan mana yang memiliki skor rata-rata tertinggi?

## Q13  Jika skala di kategorikan menjadi positif (SS dan S), netral (CS) dan negatif (CTS, TS dan STS) Maka hitung jumlah responden dan persentase masing-masing kategori

```python
import pandas as pd
import plotly.express as px

def plot_from_print(judul, hasil_print, label_x):
    skala, jumlah, persen = hasil_print.split("|")

    df_plot = pd.DataFrame({
        label_x: [skala],
        "Jumlah": [int(jumlah)],
        "Persen": [float(persen)]
    })

    fig = px.bar(
        df_plot,
        x=label_x,
        y="Jumlah",
        text="Persen",
        title=judul
    )

    fig.update_traces(texttemplate='%{text}%')
    fig.show()

hasil = "positif=1396:72.7|netral=471:24.5|negatif=54:2.8"

rows = []
for item in hasil.split("|"):
    kategori, value = item.split("=")
    jumlah, persen = value.split(":")
    rows.append([kategori.capitalize(), int(jumlah), float(persen)])

df_q13 = pd.DataFrame(rows, columns=["Kategori","Jumlah","Persen"])

fig = px.pie(
    df_q13,
    names="Kategori",
    values="Jumlah",
    title="Q13 - Distribusi Kategori Jawaban"
)

fig.show()
```
